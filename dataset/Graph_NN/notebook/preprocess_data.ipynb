```python
samples=['CRC0322_NT_1_3000','CRC0327_NT_2','CRC1502_NT_1','CRC0542_NT72h_1']
samples_KRAS=['CRC1620_NT_1','CRC1139_NT_1']
DIC_FOLDER={
 'CRC0322_cetux_1':'rCASC_Ire_cetuxi',  
 'CRC0322_NT_1_3000':'rCASC_Ire_cetuxi',
 'CRC0327_cetux_2':'rCASC_Ire_cetuxi',
 'CRC0327_NT_2':'rCASC_Ire_cetuxi',
 'CRC0069_CTX72h_1':'rCASC_longer',
 'CRC0069_NT72h_1':'rCASC_longer',
 'CRC0542_CTX72h_1':'rCASC_longer',
 'CRC0542_NT72h_1':'rCASC_longer',
 'CRC0327_NT_t1':'rCASC_IRCC_tecrep',
 'CRC0327_NT_t2':'rCASC_IRCC_tecrep',
 'CRC0327_cetux_1_4000':'rCASC_Ire_cetuxi',
 'CRC0327_NT_1':'rCASC_Ire_cetuxi',
 'CRC1502_cetux_1':'rCASC_Ire_cetuxi',
 'CRC1502_NT_1':'rCASC_Ire_cetuxi',
 }
PRJ_ROOT='/mnt/cold1/snaketree/prj/scRNA'
KRAS_PRJ_ROOT='/mnt/cold2/snaketree/prj/scRNA'
DIC_KRAS={
 'CRC1620_NT_1':'candiolo_kras_dec22_rCASC',
 'CRC1139_NT_1':'candiolo_kras_dec22_rCASC'}

```

```python
dic_path={}
#filtered_annotated_saver_ribomito_CRC0322_NT_1_3000_log2_pc1_cpm.csv

for sample in samples:
    dic_path[sample]=PRJ_ROOT+'/dataset/'+DIC_FOLDER[sample]+'/'+sample+'_dir/'+'filtered_annotated_saver_ribomito_'+sample+'_log2_pc1_cpm.csv'
for sample in samples_KRAS:
    dic_path[sample]=KRAS_PRJ_ROOT+'/dataset/'+DIC_KRAS[sample]+'/'+sample+'_dir/'+'filtered_annotated_saver_ribomito_'+sample+'_log2_pc1_cpm.csv'
```

```python
out_folder='NT_GNN_saver'
with open("gene_list_n.txt", "r") as f:
    gene_list = [line.strip() for line in f if line.strip()]
```

```python
len(gene_list)
```

```python
import pandas as pd
import os

output_dir='NT_GNN_saver'
os.makedirs(output_dir, exist_ok=True)
geni_presenti={}

for sample_name, file_path in dic_path.items():
    print(f"Processing {sample_name}...")

    
    df = pd.read_csv(file_path,header=0,index_col=0,sep=',').T
    df.columns = [col.split(":")[1] if ":" in col else col for col in df.columns]

    selected_cols = [col for col in df.columns if col in gene_list]
    df_filtered = df[selected_cols]
    geni_presenti[sample_name]=len(selected_cols)

    output_path = os.path.join(output_dir, f"filtered_{sample_name}.csv")
 
    df_filtered.to_csv(output_path)

print("Done!")
```
