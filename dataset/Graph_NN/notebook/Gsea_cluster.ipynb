```r
library(dplyr)
library(tidyr)

find_private_gs <- function(df, alpha = 0.05,
                            id_col = "ID", cl_col = "cluster",
                            nes_col = "NES", padj_col = "p.adjust") {
  d1 <- df %>%
    rename(ID      = all_of(id_col),
           cluster = all_of(cl_col),
           NES     = all_of(nes_col),
           padj    = all_of(padj_col)) %>%
    group_by(ID, cluster) %>%
    arrange(padj, desc(abs(NES)), .by_group = TRUE) %>%
    slice(1) %>%
    ungroup() %>%
    mutate(sig = padj < alpha,
           dir = sign(NES),
           status = case_when(sig & dir > 0 ~  1L,   # sig up
                              sig & dir < 0 ~ -1L,   # sig down
                              TRUE          ~  0L))  # non sig

  # 2) long: una riga per (ID, cluster) con status
  status_long <- d1 %>% select(ID, cluster, status)

  # 3) per ogni ID, marca i cluster "privati"
  private_long <- status_long %>%
    group_by(ID) %>%
    mutate(private = ifelse(
      status == 0, FALSE,
      sapply(seq_along(status), function(i) {
        oth <- status[-i]
        all(oth == 0 | oth == -sign(status[i]))
      })
    )) %>%
    ungroup() %>%
    filter(private) %>%
    select(ID, cluster, status)

  # 4) aggiungi NES/padj (join 1-1)
  private_tbl <- private_long %>%
    left_join(d1 %>% select(ID, cluster, NES, padj), by = c("ID","cluster")) %>%
    arrange(cluster, padj, desc(NES))

  # matrice di controllo ID × cluster (1=up, -1=down, 0=non sig)
  status_matrix <- status_long %>%
    tidyr::pivot_wider(names_from = cluster, values_from = status, values_fill = 0)

  list(
    private_table = private_tbl,
    private_by_cluster = split(private_tbl, private_tbl$cluster),
    status = status_matrix
  )
}


```

```r
res <- find_private_gs(gsea_all, alpha = 0.05)


privati<-res$private_table

```

```r
library(dplyr)
library(tidyr)
library(purrr)

find_private_gs_strict <- function(df, alpha = 0.05,
                                   direction = c("up","down"),
                                   id_col = "ID", cl_col = "cluster",
                                   nes_col = "NES", p_col = "p.adjust") {

  dir_target <- match.arg(direction)
  dir_val <- if (dir_target == "up")  1L else -1L   


  d1 <- df %>%
    rename(ID      = all_of(id_col),
           cluster = all_of(cl_col),
           NES     = all_of(nes_col),
           pval    = all_of(p_col)) %>%
    group_by(ID, cluster) %>%
    arrange(pval, desc(abs(NES)), .by_group = TRUE) %>%
    slice(1) %>%                           
    ungroup() %>%
    mutate(sig = pval < alpha,
           dir = sign(NES),
           status = case_when(sig & dir > 0 ~  1L,  
                              sig & dir < 0 ~ -1L,   
                              TRUE          ~  0L))  

  # 2) long: una riga per (ID, cluster) con status
  status_long <- d1 %>% select(ID, cluster, status)

  # 3) per ogni ID, cluster privati secondo la nuova regola
  private_long <- status_long %>%
    group_by(ID) %>%
    group_modify(function(.x, .y) {
      s  <- .x$status
      cl <- .x$cluster
      idx <- which(s == dir_val)              # candidati: segno richiesto e signif. nel cluster
      if (!length(idx)) return(tibble())
      map_dfr(idx, function(i) {
        others <- s[-i]
        is_private <- length(others) > 0 && all(others == -dir_val)  # tutti opposti e significativi
        if (is_private) tibble(cluster = cl[i]) else tibble()
      })
    }) %>%
    ungroup()

  # 4) aggiungi NES e p (join 1-1 pulito)
  private_tbl <- private_long %>%
    left_join(d1 %>% select(ID, cluster, NES, pval), by = c("ID","cluster")) %>%
    arrange(cluster, pval, desc(NES)) %>%
    rename(!!p_col := pval)

  # opzionale: matrice di controllo ID×cluster (1=up, -1=down, 0=non sig)
  status_matrix <- status_long %>%
    pivot_wider(names_from = cluster, values_from = status, values_fill = 0)

  list(
    private_table      = private_tbl,
    private_by_cluster = split(private_tbl, private_tbl$cluster),
    status             = status_matrix
  )
}
```

```r

res_up   <- find_private_gs_strict(gsea_all, alpha = 0.05, direction = "up",
                                   id_col = "ID", cl_col = "cluster",
                                   nes_col = "NES", p_col = "p.adjust")

res_down <- find_private_gs_strict(gsea_all, alpha = 0.05, direction = "down",
                                   id_col = "ID", cl_col = "cluster",
                                   nes_col = "NES", p_col = "p.adjust")


res_up$private_table


res_down$private_by_cluster[["0"]]


head(res_up$status)
```
