```python
import pandas as pd
import os

# === 1. Imposta cartella principale ===
base_dir = 'count_matrix'  # <-- Cambia con il tuo percorso

# === 2. Cicla su ogni sottocartella ===
for subfolder in os.listdir(base_dir):
    subfolder_path = os.path.join(base_dir, subfolder)
    if not os.path.isdir(subfolder_path):
        continue  # ignora se non è una cartella

    # === 3. Prendi solo i file che terminano con 'epithelial.csv' ===
    csv_files = [f for f in os.listdir(subfolder_path) if f.endswith('epithelial.csv')]
    stats_list = []

    for file_name in csv_files:
        sample_id = file_name.replace('.csv', '')
        file_path = os.path.join(subfolder_path, file_name)

        df = pd.read_csv(file_path)
        df = df.rename(columns={df.columns[0]: 'Gene'})
        df.set_index('Gene', inplace=True)

        total_cells = df.shape[1]
        total_umi = df.sum().sum()
        umi_per_cell = df.sum(axis=0).mean()
        genes_per_cell = (df > 0).sum(axis=0).mean()

        stats_list.append({
            'Sample': sample_id,
            'Cells': total_cells,
            'Total UMIs': total_umi,
            'Mean UMIs per Cell': umi_per_cell,
            'Mean Genes per Cell': genes_per_cell
        })

    if stats_list:
        # === 4. Ordina le righe per Sample ID ===
        stats_list = sorted(stats_list, key=lambda x: x['Sample'])

        # === 5. Crea DataFrame e salva con nome dinamico ===
        output_df = pd.DataFrame(stats_list)
        output_filename = f"epithelial_sample_statistics_{subfolder}.csv"
        output_path = os.path.join(subfolder_path, output_filename)
        output_df.to_csv(output_path, index=False)

        print(f"✔ Statistiche salvate in: {output_path}")
    else:
        print(f"⚠ Nessun file epithelial.csv trovato in {subfolder_path}")
```
