```r
library(reshape)
library(ez)

d <- read.table('~/ANOVA_DLL1.tsv', sep='\t', header=TRUE, stringsAsFactors = FALSE)
colnames(d) <- gsub('(\\w+)(\\d)', "\\1_\\2", colnames(d))
longf <- melt(d, id.vars="time")
longf$variable <- as.character(longf$variable)
longf$treat <- sapply(strsplit(longf$variable, '_'), function(x){x[[1]][1]})
longf$replicate <- sapply(strsplit(longf$variable, '_'), function(x){x[[2]][1]})

longf$treat <- as.factor(longf$treat)
longf$replicate <- as.factor(longf$replicate)
longf$time <- as.factor(longf$time)
```

```r
#https://stats.stackexchange.com/questions/23305/why-is-ezanova-not-giving-me-mauchly%C2%B4s-test-of-sphericity-even-though-i-belive
ezANOVA(data    = longf, 
        wid     = replicate, 
        within  = .(treat, time),
        dv      = value, type=2) 
```

```r
# nono, chiaramente
ezANOVA(data    = longf, 
        wid     = replicate, 
        between  = .(treat, time),
        dv      = value) 
```

```r

ezANOVA(data    = longf, 
        wid = replicate,
        within_full = treat,
        within = time,
        dv      = value) 
```
